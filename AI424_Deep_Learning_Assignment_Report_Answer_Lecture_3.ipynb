```python
import math

# Activation functions
def sigmoid(x):
    return 1 / (1 + math.exp(-x))

def tanh(x):
    return math.tanh(x)

# LSTM weights (from the PDF)
# Forget gate: input weight = 0.5, hidden weight = 0.1, bias = 0
# Input gate: input weight = 0.6, hidden weight = 0.2, bias = 0
# Candidate cell: input weight = 0.7, hidden weight = 0.3, bias = 0
# Output gate: input weight = 0.8, hidden weight = 0.4, bias = 0
w_f = (0.5, 0.1)
w_i = (0.6, 0.2)
w_c = (0.7, 0.3)
w_o = (0.8, 0.4)
bias = 0.0

# Initial states
h = 0.0
C = 0.0

# Input sequence (including the "next" value to compute h4)
X = [1, 2, 3, 4]

print("Step-by-step LSTM computation:")
for t, x in enumerate(X, start=1):
    # Forget gate
    f = sigmoid(w_f[0] * x + w_f[1] * h + bias)
    # Input gate
    i = sigmoid(w_i[0] * x + w_i[1] * h + bias)
    # Candidate cell state
    C_tilde = tanh(w_c[0] * x + w_c[1] * h + bias)
    # Cell state update
    C = f * C + i * C_tilde
    # Output gate
    o = sigmoid(w_o[0] * x + w_o[1] * h + bias)
    # Hidden state update
    h = o * tanh(C)

    print(f"Time step t={t}, x={x}")
    print(f"  f = {f:.3f}")
    print(f"  i = {i:.3f}")
    print(f"  C~ = {C_tilde:.3f}")
    print(f"  C = {C:.3f}")
    print(f"  o = {o:.3f}")
    print(f"  h = {h:.3f}\n")

# Linear output layer (assumed from PDF: weight = 4, bias = 0)
W_y = 4.0
b_y = 0.0
prediction = W_y * h + b_y
print(f"Final prediction from h4: {prediction:.3f} ≈ 3.8 (as in PDF)")
```
